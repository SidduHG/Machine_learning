# Classify a real-world table

A small historical wine dataset contains chemical measurements and three cultivar classes. Build a classifier and explain which classes it confuses. Your focus is a transparent pipeline and evaluation, not a claim that this small dataset represents all modern wine production.

Dataset: scikit-learn wine benchmark · 178 observations, 13 chemical measurements

Reference: https://scikit-learn.org/stable/datasets/toy_dataset.html#wine-recognition-dataset

## Plan

1. Inspect class balance and identify which numeric features have different scales.
2. Create a stratified 80/20 train/test split with random_state=42.
3. Fit a most-frequent baseline and a scaling + logistic-regression pipeline.
4. Report accuracy, macro F1 and a confusion matrix with class labels.
5. Inspect errors and discuss sampling limitations before proposing a serving contract.


```python
from sklearn.datasets import load_wine
from sklearn.model_selection import train_test_split
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.dummy import DummyClassifier
from sklearn.metrics import accuracy_score, f1_score, confusion_matrix

data = load_wine()
X_train, X_test, y_train, y_test = train_test_split(data.data, data.target, test_size=.2, random_state=42, stratify=data.target)
models = {'Baseline':DummyClassifier(strategy='most_frequent'), 'Logistic':make_pipeline(StandardScaler(),LogisticRegression(max_iter=1000))}
print('Classes:', list(data.target_names))
for name,model in models.items():
    model.fit(X_train,y_train)
    predictions=model.predict(X_test)
    print(name, 'Accuracy:',round(accuracy_score(y_test,predictions),3),'Macro F1:',round(f1_score(y_test,predictions,average='macro'),3))
    print(confusion_matrix(y_test,predictions))
```
