# Recognize handwritten digits

Build a small image classifier that predicts digits 0–9. Begin with logistic regression, then compare a multilayer perceptron. The network is a compact CPU-friendly scikit-learn model; extending it to a PyTorch CNN is a separate challenge.

Dataset: scikit-learn digits benchmark · 1,797 images with 64 pixel features

Reference: https://scikit-learn.org/stable/datasets/toy_dataset.html#optical-recognition-of-handwritten-digits-dataset

Original ML Atlas notebook, MIT-licensed code and CC BY 4.0 explanation. Third-party data retains its own license.


## Environment

Run in Jupyter or Colab. For the scikit-learn projects, install scikit-learn >=1.4 and its dependencies in your own environment. No GPU is needed.


## Plan

1. Load the 8×8 images, inspect pixel range, and understand the flattened 64-feature representation.
2. Make a stratified train/test split before selecting architecture.
3. Fit scaled logistic regression and a one-hidden-layer MLP with early stopping using an internal training-validation split.
4. Report test accuracy and a confusion matrix. Inspect wrong predictions by reshaping the feature vectors back into images.
5. Document training duration, parameter choices and the limits of performance on this small benchmark.


```python
from sklearn.datasets import load_digits
from sklearn.model_selection import train_test_split
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.neural_network import MLPClassifier
from sklearn.metrics import accuracy_score, confusion_matrix

digits=load_digits()
X_train,X_test,y_train,y_test=train_test_split(digits.data,digits.target,test_size=.2,stratify=digits.target,random_state=42)
models={'Linear':make_pipeline(StandardScaler(),LogisticRegression(max_iter=1000)), 'MLP':make_pipeline(StandardScaler(),MLPClassifier(hidden_layer_sizes=(48,),early_stopping=True,max_iter=120,random_state=42))}
for name,model in models.items():
    model.fit(X_train,y_train)
    predictions=model.predict(X_test)
    print(name,'Accuracy:',round(accuracy_score(y_test,predictions),4))
    print('First mistakes:',[(int(a),int(b)) for a,b in zip(y_test,predictions) if a!=b][:10])
    print(confusion_matrix(y_test,predictions))
```

- [ ] Misclassifications are inspected, not just counted.
- [ ] The notebook runs on a CPU with fixed random seeds.
