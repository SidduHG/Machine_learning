# Build an evidence-first study assistant

Build the retrieval layer for a question-answering system. Begin with a transparent lexical baseline, preserve source IDs, and refuse to invent evidence for unrelated questions. The runnable notebook performs retrieval; adding a generative model is an optional extension that requires your own provider credentials.

Dataset: A small original course-note collection included in the notebook

Reference: https://arxiv.org/abs/2005.11401

Original ML Atlas notebook, MIT-licensed code and CC BY 4.0 explanation. Third-party data retains its own license.


## Environment

Run in Jupyter or Colab. For the scikit-learn projects, install scikit-learn >=1.4 and its dependencies in your own environment. No GPU is needed.


## Plan

1. Write a small permitted collection with stable source IDs and topic-specific passages.
2. Tokenize consistently and score term overlap as a baseline.
3. Return ranked evidence and abstain when no content terms match.
4. Create a held-out set of questions with expected sources and report top-1 source accuracy.
5. Document failures from synonyms and vocabulary mismatch. Compare a TF-IDF or embedding retriever before adding generation.


```python
import re
notes={
 'gradient-descent':'gradient descent updates weights in the negative gradient direction with a learning rate',
 'regularization':'regularization penalizes large weights and can reduce overfitting',
 'pca':'principal component analysis pca projects data onto directions of high variance',
 'evaluation':'precision counts correct positive predictions recall counts recovered actual positives'
}
stop={'the','a','an','is','in','of','how','what','does','to','and','can'}
def tokens(text):
    return set(re.findall(r'[a-z]+',text.lower()))-stop
def retrieve(question):
    ranked=sorted(((len(tokens(question)&tokens(text)),key) for key,text in notes.items()),reverse=True)
    return ranked[0][1] if ranked[0][0]>0 else None
questions=[('How does gradient descent use learning rate?','gradient-descent'),('What reduces overfitting?','regularization'),('How does PCA preserve variance?','pca'),('How is recall calculated?','evaluation')]
correct=0
for question,expected in questions:
    source=retrieve(question)
    correct += source==expected
    print(question,'=>',source)
print('Top-1 source accuracy:',correct/len(questions))
print('Unrelated question:',retrieve('How should I bake a cake?'))
```
